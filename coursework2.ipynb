```python
# import used to read in the csv file
import csv

class RecordOfBike:
    def __init__(self, row, record_type):

        # These four lines are used to get information from those specific rows
        self.date = row[1]
        self.season = int(row[2])
        self.year = 2011 + int(row[3])    
        self.month = int(row[4])

        # If record type is hourly data then this will aim to get read from the hour and move the column by 1
        if record_type == "hourly":
            self.hour = int(row[5])
            offset = 1
        else:
            self.hour = None
            offset = 0

        # This gets the number of casual users, registered users and total rentals from the data
        self.casual_users = int(row[13 + offset])
        self.registered_users = int(row[14 + offset])
        self.total_rentals = int(row[15 + offset])
```

```python
class BikeDataset:

    # made for when it comes to outputting the months instead of it being numbers so it is well known
    MONTHS = {
        1:"January",2:"February",3:"March",4:"April",
        5:"May",6:"June",7:"July",8:"August",
        9:"September",10:"October",11:"November",12:"December"
    }

    def __init__(self, filename, record_type):
        # stores the filename and record type
        self.filename = filename
        self.record_type = record_type
        self.records = []
        self.load_data()  

    def load_data(self):
        # opens the file and reads each line into a RecordOfBike object
        with open(self.filename) as f:
            next(f)  # skip header line
            for line in f:
                row = line.strip().split(",")
                # turns each row from the file into a bike record and adds it to the list
                self.records.append(RecordOfBike(row, self.record_type))
        

    # this is used to get all records for a certain year 
    def get_by_year(self, year):
        found = []
        for row_data in self.records:
            if row_data.year == year:
                found.append(row_data)
        return found

    # this is used to get all records for a certain year and month
    def get_by_year_month(self, year, month):
        found = []
        for row_data in self.records:
            
            if row_data.year == year and row_data.month == month:
                found.append(row_data)
        return found

    # this is used to get all records for a certain season
    def get_by_season(self, season):
        found = []
        for row_data in self.records:
            if row_data.season == season:
                found.append(row_data)
        return found

    # works out the average rentals for each month in 2011 and 2012 and prints the difference
    def display_monthly_differences(self):
        print("\nMonthly Differences (2012 - 2011):")

        monthly = {}

        # groups all rental numbers by year, month
        for row_data in self.records:
            info_pair = (row_data.year, row_data.month)
            if info_pair not in monthly:
                monthly[info_pair] = []
            monthly[info_pair].append(row_data.total_rentals)

        # goes through each month and compares the averages for 2011 and 2012
        for month in range(1, 13):
            value_11 = monthly.get((2011, month), [])
            value_12 = monthly.get((2012, month), [])

            
            avg_2011 = sum(value_11)/len(value_11) if value_11 else 0
            avg_2012 = sum(value_12)/len(value_12) if value_12 else 0

            difference = avg_2012 - avg_2011
            print(self.MONTHS[month] + ": " + str(round(difference, 2)))

    # finds which season had the highest total rentals for each year
    def best_season(self):
        print("\nBest Season Per Year:")
        data_totals = {}

        # adds up total rentals for each year, season
        for row_data in self.records:
            info_pair = (row_data.year, row_data.season)
            if info_pair not in data_totals:
                data_totals[info_pair] = 0
            data_totals[info_pair] += row_data.total_rentals

        # checks all four seasons and finds the highest one for each year
        for year in [2011, 2012]:
            best_season = 1
            best_value = data_totals.get((year, 1), 0)

            for season in [2, 3, 4]:
                season_total = data_totals.get((year, season), 0)
                if season_total > best_value:
                    best_value = season_total
                    best_season = season

            print(str(year) + ": Season " + str(best_season))

    # finds the 3 months with the lowest total rentals for each year
    def lowest_three_months(self):
        print("\nLowest 3 Months Per Year:")
        data_totals = {}

        # adds up total rentals for each year, month
        for row_data in self.records:
            info_pair = (row_data.year, row_data.month)
            if info_pair not in data_totals:
                data_totals[info_pair] = 0
            data_totals[info_pair] += row_data.total_rentals

        # builds a list of all months and their totals for each year
        for year in [2011, 2012]:
            month_info = []
            for month in range(1, 13):
                month_info.append([month, data_totals.get((year, month), 0)])

            
            for i in range(len(month_info)):
                for j in range(i + 1, len(month_info)):
                    if month_info[j][1] < month_info[i][1]:
                        month_info[i], month_info[j] = month_info[j], month_info[i]

            print(year)
            for month, total in month_info[:3]:
                print(self.MONTHS[month] + ": " + str(total))

    # finds the 3 months with the highest total rentals for each year
    def highest_three_months(self):
        print("\nHighest 3 Months Per Year:")
        data_totals = {}

        # adds up total rentals for each year, month
        for row_data in self.records:
            info_pair = (row_data.year, row_data.month)
            if info_pair not in data_totals:
                data_totals[info_pair] = 0
            data_totals[info_pair] += row_data.total_rentals

        # builds a list of all months and their totals for each year
        for year in [2011, 2012]:
            month_info = []
            for month in range(1, 13):
                month_info.append([month, data_totals.get((year, month), 0)])

           
            for i in range(len(month_info)):
                for j in range(i + 1, len(month_info)):
                    if month_info[j][1] > month_info[i][1]:
                        month_info[i], month_info[j] = month_info[j], month_info[i]

            print(year)
            for month, total in month_info[:3]:
                print(self.MONTHS[month] + ": " + str(total))

    # finds the top 3 months with the highest casual users
    def top_three_casual(self, year):
        print("\nTop 3 Casual Months (" + str(year) + "):")
        data_totals = {}

        # adds up casual users for each month in the chosen year
        for row_data in self.records:
            if row_data.year == year:
                if row_data.month not in data_totals:
                    data_totals[row_data.month] = 0
                data_totals[row_data.month] += row_data.casual_users

        month_info = []
        for month in data_totals:
            month_info.append([month, data_totals[month]])

        
        for i in range(len(month_info)):
            for j in range(i + 1, len(month_info)):
                if month_info[j][1] > month_info[i][1]:
                    month_info[i], month_info[j] = month_info[j], month_info[i]

        for month, total in month_info[:3]:
            print(self.MONTHS[month] + ": " + str(total))

    # finds the top 3 months with the highest registered users
    def top_three_registered(self, year):
        print("\nTop 3 Registered Months (" + str(year) + "):")
        data_totals = {}

        # same as casual users above but this time using registered users instead
        for row_data in self.records:
            if row_data.year == year:
                if row_data.month not in data_totals:
                    data_totals[row_data.month] = 0
                data_totals[row_data.month] += row_data.registered_users

        month_info = []
        for month in data_totals:
            month_info.append([month, data_totals[month]])

       
        for i in range(len(month_info)):
            for j in range(i + 1, len(month_info)):
                if month_info[j][1] > month_info[i][1]:
                    month_info[i], month_info[j] = month_info[j], month_info[i]

        for month, total in month_info[:3]:
            print(self.MONTHS[month] + ": " + str(total))

```

```python
# function used to load correct dataset
def load_dataset(record_type):
    if record_type.lower() == "daily":
        return BikeDataset("day.csv", "daily")
    elif record_type.lower() == "hourly":
        return BikeDataset("hour.csv", "hourly")
    else:
        raise ValueError("record_type must be 'daily' or 'hourly'")


# running the daily dataset
daily = load_dataset("daily")

daily.display_monthly_differences()
daily.best_season()
daily.lowest_three_months()
daily.highest_three_months()

daily.top_three_casual(2011)
daily.top_three_registered(2011)
daily.top_three_casual(2012)
daily.top_three_registered(2012)
```
